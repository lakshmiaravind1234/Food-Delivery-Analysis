#creating A Box plot

```python
num_cols = df.select_dtypes(include="number").columns
```

```python
num_cols
```

```python
#Box plot
for i in num_cols:
  sns.boxplot(df[i])
  plt.title(f"Box plot of {i}")
  plt.show()
```

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
df['Time_Taken_Min']=df['Time_Taken_Min'].astype(object)
```

```python
df.dtypes
```

```python
num_colls = df.select_dtypes(include = "number").columns
```

```python
df[num_colls] = df[num_colls].fillna(df[num_colls].median())
```

```python
df.isnull().sum()
```

```python
df['Time_Taken_Min']=df['Time_Taken_Min'].astype(float)
```

```python
df.dtypes
```

```python
cat_cols = df.select_dtypes(include = "object").columns
```

```python
cat_cols
```

```python
df[cat_cols] = df[cat_cols].fillna(df[cat_cols].mode().iloc[0])
```

```python
df.isnull().sum()
```

```python
#Outlier
Q3 = df['Time_Taken_Min'].quantile(0.75)
Q1 = df['Time_Taken_Min'].quantile(0.25)
IQR = Q3 - Q1
IQR
```

```python
low_limit = Q1 - 1.5*IQR
upp_limit = Q3 + 1.5*IQR
low_limit , upp_limit
```

```python
df = df[(df['Time_Taken_Min'] >=low_limit) & (df['Time_Taken_Min']<=upp_limit)]
```

```python
df.shape
```

```python
#histogram
plt.figure(figsize=(10,6))
sns.histplot(data=df,x="Time_Taken_Min",bins=30,kde=True)
plt.title("Distribution of Delivery Time")
plt.xlabel("Delivery Time (Minutes)")
plt.ylabel("Number of Orders")
plt.show()
```

```python
#groupby
city_time = (df.groupby("City")["Time_Taken_Min"].mean().sort_values(ascending=False).head(10))
plt.figure(figsize=(10,6))
city_time.plot(kind="bar")
plt.title("Top 10 Cities by Average Delivery Time")
plt.xlabel("City")
plt.ylabel("Average Time (Minutes)")
plt.xticks(rotation=45)
plt.show()
```

```python
#pie chart
festival_count = df["Festival"].value_counts()
plt.figure(figsize=(7,7))
plt.pie(festival_count.values,labels=festival_count.index,autopct="%1.1f%%",startangle=90)
plt.title("Festival vs Non-Festival Orders")
plt.show()
```

```python
# line chart
hourly_time = (df.groupby("Order_Hour")["Time_Taken_Min"].mean().sort_index())
plt.figure(figsize=(10,6))
plt.plot(hourly_time.index,hourly_time.values,marker="o")
plt.title("Average Delivery Time by Order Hour")
plt.xlabel("Order Hour")
plt.ylabel("Average Delivery Time (Minutes)")
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(10,6))
corr = df.select_dtypes(include="number").corr()
sns.heatmap(corr,annot=True,fmt=".2f",cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()
```
